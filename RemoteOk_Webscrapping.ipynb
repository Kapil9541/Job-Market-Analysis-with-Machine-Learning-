```python
# Start browser
driver = webdriver.Firefox(
    
    service=Service(
        GeckoDriverManager().install()
    )
)

# Store data
data = []


# More categories
categories = [

    "remote-data-jobs",
    "remote-python-jobs",
    "remote-dev-jobs",
    "remote-analyst-jobs",
    "remote-machine-learning-jobs",
    "remote-backend-jobs",
    "remote-frontend-jobs",
    "remote-full-stack-jobs",
    "remote-software-jobs",
    "remote-engineer-jobs",
    "remote-javascript-jobs",
    "remote-react-jobs",
    "remote-java-jobs",
    "remote-ai-jobs",
    "remote-cloud-jobs",
    "remote-sql-jobs",
    "remote-product-manager-jobs",
    "remote-marketing-jobs",
    "remote-finance-jobs",
    "remote-design-jobs"
]


# Loop through categories
for category in categories:
    
    print(f"\nScraping Category: {category}")
    
    
    # More pages
    for page in range(1, 6):
        
        try:
            
            url = f"https://remoteok.com/{category}?page={page}"
            
            print(f"Page: {page}")
            
            driver.get(url)
            
            time.sleep(5)
            
            
            soup = BeautifulSoup(
                driver.page_source,
                "html.parser"
            )
            
            
            jobs = soup.find_all(
                "tr",
                class_="job"
            )
            
            print(f"Jobs Found: {len(jobs)}")
            
            
            for job in jobs:
                
                
                # Job title
                try:
                    title = job.find("h2").text.strip()
                except:
                    title = "N/A"
                
                
                # Company
                try:
                    company = job.find("h3").text.strip()
                except:
                    company = "N/A"
                
                
                # Location
                try:
                    location = job.find(
                        "div",
                        class_="location"
                    ).text.strip()
                except:
                    location = "Worldwide"
                
                
                # Skills / Tags
                try:
                    
                    tags = job.find_all(
                        "div",
                        class_="tag"
                    )
                    
                    skills = [
                        tag.text.strip()
                        for tag in tags
                    ]
                    
                    skills = ", ".join(skills)
                    
                except:
                    skills = "N/A"
                
                
                # Salary
                try:
                    salary = job.find(
                        "div",
                        class_="salary"
                    ).text.strip()
                except:
                    salary = "Not Mentioned"
                
                
                # Date Posted
                try:
                    date_posted = job.find(
                        "time"
                    ).text.strip()
                except:
                    date_posted = "N/A"
                
                
                # Job Link
                try:
                    
                    link = job.get("data-href")
                    
                    job_link = "https://remoteok.com" + link
                    
                except:
                    job_link = "N/A"
                
                
                # Description
                try:
                    description = job.find(
                        "td",
                        class_="description"
                    ).text.strip()
                except:
                    description = "N/A"
                
                
                # Save row
                data.append([
                    
                    category,
                    title,
                    company,
                    location,
                    skills,
                    salary,
                    date_posted,
                    job_link,
                    description
                ])
        
        
        except Exception as e:
            
            print("Error:", e)


# Close browser
driver.quit()


# Create DataFrame
df = pd.DataFrame(
    
    data,
    
    columns=[
        
        "Category",
        "Job_Title",
        "Company",
        "Location",
        "Skills",
        "Salary",
        "Date_Posted",
        "Job_Link",
        "Description"
    ]
)


# IMPORTANT:
# Comment this initially
# to check raw row count

# df.drop_duplicates(inplace=True)


# Show shape
print("\nDataset Shape:")
print(df.shape)


# Display first rows
print(df.head())


# Save CSV
df.to_csv(
    "remoteok_1000plus_jobs.csv",
    index=False
)
```
